```python

import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
from google.colab import files

# Upload dataset
uploaded = files.upload()
df = pd.read_csv(next(iter(uploaded)))

# Display first five rows
print("First five rows:")
print(df.head())

# Shape of dataset
print("\nShape of Dataset:")
print(df.shape)

# Column names
print("\nColumn Names:")
print(df.columns)

# Dataset information
print("\nDataset Information:")
df.info()

# Statistical summary
print("\nStatistical Summary:")
print(df.describe())

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Mean imputation
df_mean = df.copy()
mean = df_mean["BuildingArea"].mean()
df_mean["BuildingArea"] = df_mean["BuildingArea"].fillna(mean)

print("\nMean Imputation:")
print(df_mean["BuildingArea"].isnull().sum())

# Median imputation
df_median = df.copy()
median = df_median["YearBuilt"].median()
df_median["YearBuilt"] = df_median["YearBuilt"].fillna(median)

print("\nMedian Imputation:")
print(df_median["YearBuilt"].isnull().sum())

# Forward fill
df_ffill = df.copy()
df_ffill["Car"] = df_ffill["Car"].ffill()

print("\nForward Fill:")
print(df_ffill["Car"].isnull().sum())

# Mode imputation
df_processed = df.copy()
mode = df_processed["CouncilArea"].mode()[0]
df_processed["CouncilArea"] = df_processed["CouncilArea"].fillna(mode)

print("\nMode Imputation:")
print(df_processed["CouncilArea"].isnull().sum())

# One-Hot Encoding
df_encoded = pd.get_dummies(
    df_processed,
    columns=["Type", "Method"],
    dtype=int
)

print("\nOne-Hot Encoding:")
print(df_encoded.head())

# Label Encoding
df_label = df.copy()
le = LabelEncoder()

df_label["CouncilArea_Encoded"] = le.fit_transform(
    df_label["CouncilArea"].fillna("Unknown")
)

print("\nLabel Encoding:")
print(df_label[["CouncilArea", "CouncilArea_Encoded"]].head(10))

```

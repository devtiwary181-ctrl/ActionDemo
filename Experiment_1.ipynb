```python
import numpy as np
import pandas as pd
```

```python
arr = np.array([10,20,30,40,50])
print("\nn1. NumPy Array")
print("Array:",arr)
```

```python
print("First element:",arr[0])
print("Third element:",arr[2])
print("Last element:",arr[-1])
```

```python
print("\n--- Array Slicing ---")
print("First three elements:",arr[0:3])
print("Elements from index 2:",arr[2:])
print("Last three elements:",arr[-3:])
```

```python
print("\n--- Basic Array Operations ---")
print("Add 5:",arr + 5)
print("Subtract 5:",arr - 5)
print("Multiply by 2:",arr*2)
print("Divide by 10:",arr/10)
```

```python
print("\n--- Mathematical Operations ---")
print("Sum:",np.sum(arr))
print("Mean:",np.mean(arr))
print("Maximum:",np.max(arr))
print("Minimum:",np.min(arr))
```

```python
arr=np.array([10,20,30,40,50,60])
new_arr=arr.reshape(6,1)
print("Original Array:")
print(arr)
print("\nReshaped Array:")
print(new_arr)
```

```python
data = {
    "Name": ["Asha","Ravi","Meera","Karan"],
    "Marks":[88,72,91,65],
    "Branch":["CSE","ECE","CSE","ISE"]
}
df=pd.DataFrame(data)
print("\n--- Pandas DataFrame ---")
print(df)
```

```python
print(df[["Name","Marks"]])
```

```python
print("Average Marks:",df["Marks"].mean())
print("Highest Marks:",df["Marks"].max())
print("Lowest Marks:",df["Marks"].min())
```

```python
print("--- Original DataFrame ---")
print(df)
print("\nFirst Row:")
print(df.iloc[0])
print("\nSecond Row:")
print(df.iloc[1])
print("\nThird Row:")
print(df.iloc[2])
print("\nFourth Row:")
print(df.iloc[3])
```

```python
print("\n--- Specific Values ---")
print("Marks of first student:",df.iloc[0,1])
print("Marks of Ravi:",df.iloc[1,1])
print("Branch of Meera:",df.iloc[2,2])
print("Name of Karan:",df.iloc[3,0])
```

```python
print("\n--- Data Filtering ---")
result =df[df["Marks"]>85]
print("Students with Marks greater than 85:")
print(result)
```
